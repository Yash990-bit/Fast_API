# HTTP Methods

HTTP Methods batati hain ki client (browser/app) server ke saath kya action perform karna chahta hai.

# 1. GET

Data fetch/retrieve karne ke liye

# Use

✅ Read Data

❌ Data modify nahi karta

# 2. POST

Naya data create karne ke liye

# Use

✅ Create New Record

# 3. PUT

Pura resource update karne ke liye

# Use

✅ Complete Update

# 4. PATCH

Partial update ke liye

# Use

✅ Partial Update

# 5. DELETE

Data delete karne ke liye

# Use

✅ Delete Record


# Quick Memory Trick
GET    → Read

POST   → Create

PUT    → Full Update

PATCH  → Partial Update

DELETE → Delete
