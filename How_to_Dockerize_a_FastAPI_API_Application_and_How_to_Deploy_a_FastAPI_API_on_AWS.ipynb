# How to Deploy a FastAPI API on AWS

The most common beginner-friendly approach is:

FastAPI App
    ↓
Dockerize App
    ↓
Push Docker Image
    ↓
AWS EC2
    ↓
Run Docker Container
    ↓
Public API URL



# Step 1: Create EC2 Instance

Login to AWS Console

Open Amazon EC2

Click Launch Instance

# Select:
Ubuntu Server 24.04 LTS

t2.micro (Free Tier eligible)

Create/Download Key Pair (.pem)

# Allow:
SSH (22)

HTTP (80)

HTTPS (443)

Custom TCP (8000)

Launch instance.

# Step 2: Connect to EC2

ssh -i mykey.pem ubuntu@<public-ip>

* Example:

ssh -i aws.pem ubuntu@13.126.xx.xx

# Step 3: Install Docker

Update system:

sudo apt update

Install Docker:

sudo apt install docker.io -y

Start Docker:

sudo systemctl start docker

sudo systemctl enable docker

Verify:

docker --version

# Step 4: Build FastAPI Docker Image Locally

docker build -t fastapi-app .

Test locally:

docker run -p 8000:8000 fastapi-app

# Step 5: Push Image to Docker Hub

Create account on:

Docker Hub

Login:

docker login

Tag image:

docker tag fastapi-app yourusername/

fastapi-app:latest

Push:

docker push yourusername/fastapi-app:latest

# Step 6: Pull Image on EC2

SSH into EC2:

docker pull yourusername/fastapi-app:latest

# Step 7: Run Container
docker run -d \

-p 8000:8000 \

--name fastapi-api \

yourusername/fastapi-app:latest

Check:

docker ps

# Step 8: Configure Security Group

Add inbound rule:

Type	Port

Custom TCP	8000

Save changes.

# Step 9: Access API
http://<EC2-PUBLIC-IP>:8000

Swagger Docs:

http://<EC2-PUBLIC-IP>:8000/docs

OpenAPI:

http://<EC2-PUBLIC-IP>:8000/openapi.json
