# Update Endpoint

Update endpoint ka use existing data ko update karne ke liye hota hai.

Usually:

PUT → Complete update

PATCH → Partial update

# PUT Endpoint Code

```python
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI()

class User(BaseModel):
    name: str
    age: int

@app.put("/users/{user_id}")
def update_user(user_id: int, user: User):
    return {
        "user_id": user_id,
        "updated_user": user
    }
```

```python
# Request
PUT /users/1
```

```python
# Body:

{
  "name": "Yash",
  "age": 22
}
```

```python
# Response
{
  "user_id": 1,
  "updated_user": {
    "name": "Yash",
    "age": 22
  }
}
```

# How It Works

@app.put("/users/{user_id}")

# ➡️ Update endpoint

user_id: int

# ➡️ Path parameter (which user to update)

user: User

# ➡️ Request body (new data)

# Delete
DELETE method ka use server se data/resource delete karne ke liye hota hai.



# DELETE Code



```python
from fastapi import FastAPI

app = FastAPI()

@app.delete("/users/{user_id}")
def delete_user(user_id: int):
    return {"message": f"User {user_id} deleted"}
```

```python
# Request

DELETE /users/1

Yahan:

1
↑
Path Parameter

Batata hai kaunsa user delete karna hai.

# Response
{
    "message": "User 1 deleted"
}
```

# How It Works

@app.delete("/users/{user_id}")

# ➡️ DELETE endpoint create karta hai.

user_id: int

# ➡️ URL se user ID leta hai.

return {"message": ...}
