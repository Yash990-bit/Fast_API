# Path Parameters

Path Parameters URL ke andar values pass karne ka tarika hain.

# Example URL
/users/101

Yahan 101 ek path parameter hai.

@app.get("/users/{user_id}")
def get_user(user_id: int):
    return {"user_id": user_id}
Request

GET /users/101

Response
{
  "user_id": 101
}

FastAPI automatically URL se 101 ko user_id variable mein store kar deta hai.

```python
# Multiple Path Parameters
@app.get("/users/{user_id}/orders/{order_id}")
def get_order(user_id: int, order_id: int):
    return {
        "user_id": user_id,
        "order_id": order_id
    }
```

Request

GET /users/101/orders/500

Response
{
  "user_id": 101,
  "order_id": 500
}

# Path Function

FastAPI mein Path Function woh function hota hai jo kisi specific URL (path) par request aane par execute hota hai



# Example

```python
from fastapi import FastAPI

app = FastAPI()

@app.get("/")
def home():
    return {"message": "Hello World"}
```

Yahan:

def home():

ye Path Function hai.

Aur

@app.get("/")

ye batata hai ki / path par request aaye to home() function chale.

# Query Parameters (KEY-VALUE Pairs)

Query Parameters URL mein ? ke baad aane wali values hoti hain.

# Example
/users?user_id=101

Yahan:

user_id=101

ek query parameter hai.


# Example

from fastapi import FastAPI

app = FastAPI()

@app.get("/users")

def get_user(user_id: int):

    return {"user_id": user_id}

Request

GET /users?user_id=101

Response

{

  "user_id": 101

}

# commonly used for filtering, searching, sorting, and pagination. In FastAPI, they are automatically extracted from the URL and validated using type hints. multiple parameter are separeted by (**&**)

```python
# Query Parameter

from fastapi import FastAPI

app = FastAPI()

@app.get("/users")
def get_user(user_id: int):
    return {"user_id": user_id}
```

```python
# Request
GET /users?user_id=101

# Response
{
  "user_id": 101
}
```

```python
# Path + Query Parameter

from fastapi import FastAPI

app = FastAPI()

@app.get("/users/{user_id}")
def get_user(user_id: int, name: str):
    return {
        "user_id": user_id,
        "name": name
    }
#Request
GET /users/101?name=Yash

#Response

{
  "user_id": 101,
  "name": "Yash"
}
```

# Easy Memory Trick
# Path Parameter = Identity Card
/users/101
       ↑
   User ID

Batata hai exactly kaunsa resource chahiye.

# Query Parameter = Filter
/users?city=Delhi
       ↑
     Filter

Batata hai resource ko kaise search/filter karna hai.



# Path parameters are used to identify a specific resource and are part of the URL path, while query parameters are used to filter, search, sort, or paginate data and appear after the ? in the URL.
