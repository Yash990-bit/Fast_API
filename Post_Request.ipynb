# Problem Without Pydantic

Suppose user sends:

{
  "name": "Yash",
  "age": "twenty"
}

Aap expect kar rahe ho:

name: str

age: int

Without validation, application crash ya incorrect data store kar sakti hai.

# Solution: Pydantic
from pydantic import BaseModel

class User(BaseModel):
    name: str
    age: int

FastAPI automatically check karega:

name string hai?

age integer hai?

Agar nahi hai to error return karega.

# Pydantic is used in FastAPI for data validation, parsing, and type checking. It uses Python type hints to automatically validate incoming request data, reduce boilerplate code, prevent invalid data from entering the application, and help FastAPI generate API documentation automatically

# Code


```python
# Import Libraries

from fastapi import FastAPI
from pydantic import BaseModel
```

```python
 #Create FastAPI App
app = FastAPI()
```

```python
# Create Pydantic Model (Request Body)
class User(BaseModel):
    name: str
    age: int
    email: str
```

```python
# Create POST Endpoint
@app.post("/users")
def create_user(user: User):
    return {
        "message": "User created successfully",
        "user": user
    }
```
