# Step 1: What is an API?

Suppose you have a food delivery app.

You press:

"Order Pizza"

Does the app cook the pizza?

❌ No.

It sends a request to the restaurant.

Restaurant prepares the pizza and sends it back.

You
   │
   │ Request
   ▼
Restaurant
   │
   │ Response
   ▼
You

In software,

You = Client (Frontend)

Restaurant = Backend Server

Order = API Request

So,

API is a messenger between frontend and backend.

```python
# Step 2: What is FastAPI?

FastAPI is a Python framework used to create APIs.

Example:

from fastapi import FastAPI

app = FastAPI()

@app.get("/")
def home():
    return {"message": "Hello"}

This creates one API.
```

# Step 3: How do we use this API?

Run:

uvicorn main:app --reload

Now server starts.

Server Running

http://127.0.0.1:8000

```python
# Step 4: How can we test this API?

Option 1:

Browser

http://127.0.0.1:8000/

Output

{
  "message": "Hello"
}

Works!

But what if you have...

100 APIs?

Like

/users

/products

/orders

/login

/payment

/profile

/cart

Will you remember all?

❌ No.
```

# Step 5: The Problem

Suppose someone gives you a backend.

You don't know

available APIs

input

output

request body

You'll keep asking:

"Bhai API ka naam kya hai?"

"Body me kya bhejna hai?"

"Response kya aayega?"

This is difficult.

# Step 6: Swagger UI solves this problem

Swagger UI automatically creates a webpage that shows all APIs.

Think of it like this:

Backend
      │
      │
      ▼
Swagger UI

It reads your FastAPI code and builds a documentation page.

# Step 7: Where is Swagger UI?

FastAPI gives it automatically.

Open

http://127.0.0.1:8000/docs

You'll see a page like:

GET /

POST /users

GET /users/{id}

DELETE /users/{id}

Everything is listed.

# Step 8: Why is it called UI?

UI means User Interface.

Instead of typing commands in a terminal,

you get buttons like

GET /

▼

Try it out

Execute

You can click them.

No coding required.

```python
# Step 9: What happens when you click "Try it out"?

Suppose API

@app.get("/square/{num}")
def square(num: int):
    return {"square": num * num}

Swagger UI shows

GET /square/{num}

Click

Try it out

Enter

5

Click

Execute

It sends this request to your backend.

Browser
      │
      │ num = 5
      ▼
FastAPI Server

Server calculates

5 × 5 = 25

Returns

{
  "square": 25
}

Swagger UI displays it immediately.
```

# Step 10: Does Swagger UI create APIs?

❌ No.

It does not create APIs.

It only:

shows APIs

explains APIs

lets you test APIs

The real API is still running in FastAPI.

```python
# Final Flow
You write FastAPI code
          │
          ▼
Run Uvicorn
          │
          ▼
FastAPI Server starts
          │
          ▼
Swagger UI is generated automatically
          │
          ▼
Open http://127.0.0.1:8000/docs
          │
          ▼
See all APIs
          │
          ▼
Click "Try it out"
          │
          ▼
Send Request
          │
          ▼
Receive Response
```

# Swagger UI is an automatically generated web page in FastAPI that documents all your APIs and lets you test them directly from your browser without using tools like Postman.

*  Why is this useful?

# Without Swagger UI:

You need to remember endpoint URLs.

You need to know required parameters.

You need another tool (like Postman) to test APIs.

# With Swagger UI:

All endpoints are listed.

Required inputs are shown.

You can test requests immediately.

Responses are displayed automatically.

```python
# Complete Architecture
                You write Python code
                        │
                        ▼
                 FastAPI Framework
                        │
        ┌───────────────┼───────────────┐
        │               │               │
        ▼               ▼               ▼
     API Routes    OpenAPI JSON    Swagger UI
                        ▲               │
                        │               │
                        └──── Reads ────┘
```
