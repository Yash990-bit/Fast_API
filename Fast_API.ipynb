# What is API?

API = Application Programming Interface

API ek beech ka helper (messenger) hai jo ek app ki request dusri app ya server tak pahunchata hai aur uska response wapas laata hai.

# Restaurant Example

   ↓

   ↓

# Need for API?

API isliye chahiye taaki ek software dusre software se baat kar sake.

# API in ML (Machine Learning) Perspective

API ek rasta hai jisse hum apne Machine Learning model se baat karte hain.

# Example

Tumne ek ML model banaya jo house price predict karta hai.

Ab user input deta hai:

Area = 1200 sq ft

Bedrooms = 3

Direct model se baat nahi karega.

Process:

User
 ↓
API
 ↓
ML Model
 ↓
Prediction
 ↓
API
 ↓
User

Output:

House Price = ₹50 Lakhs

# Why API in ML?

Maan lo model server par deployed hai.

Website, mobile app, ya chatbot prediction lena chahta hai.

API bridge ka kaam karti hai.

Website
   ↓
API
   ↓
ML Model
   ↓
Prediction

# Layman Definition

ML mein API ek gate ya bridge hai jiske through applications machine learning model ko input bhejti hain aur prediction wapas leti hain.
